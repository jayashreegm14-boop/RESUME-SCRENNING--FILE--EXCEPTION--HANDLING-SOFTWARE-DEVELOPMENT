```python

from abc import ABC, abstractmethod

# =================================
# 1. ABSTRACT CLASS
# =================================
class Candidate(ABC):

    def __init__(self, name, qualification, skills):
        self.__name = name
        self.__qualification = qualification
        self.__skills = skills

    # Encapsulation
    def get_details(self):
        return self.__name, self.__qualification, self.__skills

    @abstractmethod
    def check_experience(self):
        pass


# =================================
# 2. INHERITANCE - FRESHER
# =================================
class FresherCandidate(Candidate):

    def __init__(self, name, qualification, skills):
        super().__init__(name, qualification, skills)

    def check_experience(self):
        return True


# =================================
# 3. INHERITANCE - EXPERIENCED
# =================================
class ExperiencedCandidate(Candidate):

    def __init__(self, name, qualification, skills, experience):
        super().__init__(name, qualification, skills)
        self.experience = experience

    def check_experience(self):
        return self.experience >= 2


# =================================
# 4. JOB CLASS
# =================================
class Job:

    def __init__(self, role, required_skills, min_experience):
        self.role = role
        self.required_skills = required_skills
        self.min_experience = min_experience


# =================================
# 5. RESUME SCREENING CLASS
# =================================
class ResumeScreening:

    def screen_candidate(self, candidate, job):

        name, qualification, skills = candidate.get_details()

        # Skill matching
        matched = set(skills) & set(job.required_skills)

        percentage = (len(matched) / len(job.required_skills)) * 100

        # Experience eligibility
        if isinstance(candidate, FresherCandidate):
            eligible = job.min_experience == 0
        else:
            eligible = candidate.experience >= job.min_experience

        # Polymorphism
        experience_status = candidate.check_experience()

        if percentage >= 50 and eligible and experience_status:
            status = "SHORTLISTED"
        else:
            status = "REJECTED"

        # Display result
        print("\n========== RESUME RESULT ==========")
        print("Name:", name)
        print("Qualification:", qualification)
        print("Skills:", skills)
        print("Job Role:", job.role)
        print("Matched Skills:", matched)
        print("Skill Matching:", round(percentage, 2), "%")
        print("Experience Eligible:", eligible)
        print("Screening Status:", status)


# =================================
# 6. MAIN PROGRAM
# =================================
try:

    job = Job(
        "Python Developer",
        ["python", "sql", "java", "html"],
        0
    )

    print("===== RESUME SCREENING SYSTEM =====")

    name = input("Enter candidate name: ")
    qualification = input("Enter qualification: ")

    skills = input("Enter skills (comma separated): ").lower().split(",")

    choice = int(input("1. Fresher\n2. Experienced\nEnter choice: "))

    if choice == 1:
        candidate = FresherCandidate(name, qualification, skills)

    elif choice == 2:
        experience = float(input("Enter experience in years: "))
        candidate = ExperiencedCandidate(
            name, qualification, skills, experience
        )

    else:
        raise ValueError("Invalid candidate choice!")

    screening = ResumeScreening()
    screening.screen_candidate(candidate, job)

except ValueError as e:
    print("Error:", e)

except Exception as e:
    print("Unexpected Error:", e)
```
